# XGBoost Modeling
This notebook loads preprocessed data, trains an XGBoost model, evaluates it, and inspects feature importance.

```python
import sys
from pathlib import Path

# Add the parent folder's src directory to the path
sys.path.append(str(Path("..", "src").resolve()))
```

```python
from preprocessing import load_and_preprocess_data

X, y = load_and_preprocess_data("../data/diabetic_data.csv")
print(X.shape, y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=260,
    learning_rate=0.0203,
    max_depth=10,
    min_child_weight=3,
    gamma=0.33,
    subsample=0.5899,
    colsample_bytree=0.5294,
    reg_alpha=0.3815,
    reg_lambda=1.5034,
    objective="multi:softprob",
    num_class=3,
    random_state=42,
    n_jobs=-1
)

sample_weights = y_train.map({0: 0.618, 1: 2.987, 2: 0.954})
model.fit(X_train, y_train, sample_weight=sample_weights)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import shap

explainer = shap.Explainer(model)
shap_values = explainer(X_train[:100])
shap.plots.beeswarm(shap_values)
```

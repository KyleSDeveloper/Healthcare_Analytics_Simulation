# 🧼 Data Cleaning Notebook
This notebook prepares the diabetes dataset for modeling by handling missing values, removing irrelevant features, and generating a clean target.

```python
import pandas as pd
import numpy as np

# Load the data
df = pd.read_csv("diabetic_data.csv")
df.head()
```

```python
# Drop irrelevant ID columns
df.drop(columns=['encounter_id', 'patient_nbr'], inplace=True)
```

```python
race_dummies = pd.get_dummies(df['race'], prefix='race', dummy_na=False)
print(f"\nCreated {len(race_dummies.columns)} race dummy columns:")
print(race_dummies.columns.tolist())
```

```python
# Replace "?" with actual NaN
df.replace('?', np.nan, inplace=True)
df.info()
```

```python
# Drop columns with too many missing values or unclear relevance
cols_to_drop = ['weight', 'payer_code', 'medical_specialty']
df.drop(columns=cols_to_drop, inplace=True)
```

```python
# Drop rows with missing values in critical columns
df.dropna(subset=['race', 'gender', 'age'], inplace=True)
```

```python
# Convert 'readmitted' to binary: 1 if <30 days, 0 otherwise
df['readmitted_binary'] = df['readmitted'].apply(lambda x: 1 if x == '<30' else 0)

# Drop original target
df.drop(columns=['readmitted'], inplace=True)

# Show updated class distribution
df['readmitted_binary'].value_counts()
```

```python
# Save cleaned data
df.to_csv("data/diabetic_data_cleaned.csv", index=False)
print("✅ Cleaned data saved to data/diabetic_data_cleaned.csv")
```

✅ Dataset is now cleaned and ready for encoding and modeling.
